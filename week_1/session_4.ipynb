# Lab 13: Grouping and aggregation

```python
df = pd.read_csv("orders.csv")
print(
    df
    .groupby('customer_id')
    .agg(
        total_amount=('amount', 'sum'),
        avg_amount=('amount', 'mean'),
        order_count=('amount', 'count')
    )
    .reset_index()
    .sort_values("total_amount", ascending=False, ignore_index=True)
)
```

# Final Integrated Lab: Mini ETL — Sales report from raw files

```python
customers = pd.read_csv("customers.csv")
orders = pd.read_csv("orders.csv")
products = pd.read_json("products.json")

df = (
    orders.merge(customers, on="customer_id", how =  'left')
    .merge(products[['product_name', "category"]], left_on="product", right_on="product_name", how = 'left')
    )

df['city'] = df.city.fillna('Unknown')
df['amount_with_tax'] = round(df.amount * 1.18,2)
df['category'] = df.category.fillna('Unknown')

report = (
    df.groupby(['city','category'])
    .agg(
        total_sales=('amount_with_tax','sum'),
        avg_order_value=('amount_with_tax','mean'),
        order_count=('order_id','count'))
    .sort_values(by='total_sales', ascending=False)
    .reset_index()
    )
report.to_csv("city_category_report.csv", index=False)

# Show top rows
print(report)
```
