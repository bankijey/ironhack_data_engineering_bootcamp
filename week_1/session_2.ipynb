# Lab 14: While loop simulating balance updates

```python
balance = 500.0
transactions = [100.0, -50.0, 0.0, -600.0, 200.0]
while balance > 0:
    for transaction in transactions:
        if transaction == 0:
            continue
        balance += transaction
        if balance < 0:
            print("Processing stopped early due to negative balance.")
            break

print(f"Final balance: {balance}")
```

# Lab 16: Final Integrated Lab: Orders mini-ETL

```python
from enum import unique
orders = [
    {"order_id": 1, "customer": "Asha", "amount": 250.0, "product_id": 1001, "status": "delivered"},
    {"order_id": 2, "customer": "Ravi", "amount": 100.0, "product_id": 1002, "status": "pending"},
    {"order_id": 3, "customer": "Asha", "amount": 150.0, "product_id": 1003, "status": "delivered"},
    {"order_id": 4, "customer": "Meera", "amount": 400.0, "product_id": 1001, "status": "delivered"},
    {"order_id": 5, "customer": "Ravi", "amount": 50.0, "product_id": 1002, "status": "cancelled"}
]

totals = {}
delivered_count = 0
unique_delivered = []

for order in orders:
    order_id = order["order_id"]
    customer = order["customer"]
    amount = order["amount"]
    product_id = order["product_id"]
    status = order["status"]

    if status != 'delivered':
        continue
    else:
        delivered_count += 1
        if customer not in totals:
            totals[customer] = {'total': amount}
        else:
            totals[customer]['total'] += amount
        unique_delivered.append(product_id)

print("Totals by customer:")
for name, record in totals.items():
    print(f"{name}: {record['total']}")
print('\n')
print(f"Unique products sold (delivered): {set(unique_delivered)}")
print(f"Delivered order count: {delivered_count}")

```
