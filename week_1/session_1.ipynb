# Lab 12: Marks percentage calculator (optional preview: simple if/else)

```python
m1 = "78"
m2 = "85"
m3 = "69"
total = sum([int(m1),int(m2),int(m3)])
pct  = total/3
print(f"Percentage {pct:.2f}%")
print(f"Result: Pass" if pct >= 35 else "Result: Fail")
```

# Lab 13: Discount calculation and rounding

```python
import math

original_price = 4999.99
discount_percent = 12.5

discount_amount = original_price * (discount_percent / 100)
final_price = original_price - discount_amount

print(f"Discount: {math.floor(discount_amount*100)/100}")
print(f"Final Price: {final_price:.2f}")
```
