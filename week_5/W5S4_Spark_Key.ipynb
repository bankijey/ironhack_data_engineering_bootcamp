```python
import pandas as pd
stopwords_RDD = sc.parallelize(stopwords)
RDD = sc.textFile("sample.txt")
RDD2 = RDD.flatMap(lambda x: x.split(" ")).map(lambda x: x.strip().lower()).filter(lambda x: x not in stopwords)

RDD3 = RDD2.map(lambda x: (x,1))
RDD4 = RDD3.reduceByKey(lambda x,y: x+y).sortBy(lambda pair: pair[1], ascending=False)
RDD4.collect()
```
